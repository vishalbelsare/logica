```python
%%logica E, M, G

# Example from https://www.youtube.com/live/8gGWdbRt4U8?si=nafg8y0FtwhGSNBl&t=1056

@Engine("duckdb");

M0(0);
E(i, i + 1) :- i in Range(10);

M(x) :- ~M(), M0(x);

@Recursive(M, 23);
M(j) :- M(i), E(i, j);   # Message passing.
M(i) :- M(i), ~E(i, j);  # Message retention.

# M(x) :-
#   M(i),
#   targets = List{j :- M(i, k)


G(i, j) :- E(i, j);
G(i, i) :- M(i);

```

```python
%%logica E, M, Done, MaxTime, ActiveEdges
#, G

# Example from https://www.youtube.com/live/8gGWdbRt4U8?si=nafg8y0FtwhGSNBl&t=1056

@Engine("duckdb");

M0(0);
E(i, i + 1) :- i in Range(10);

M(time: 0, location: x) :- M0(x);

@Recursive(M, ∞, stop: Done);
@OrderBy(M, "time");
M(time: n + 1, location: j) :- M(time: n, location: i), E(i, j);   # Message passing.
M(time: n + 1, location: i) :- M(time: n, location: i), ~E(i, j);  # Message retention.

MaxTime() = Max{t :- M(time: t)};

ActiveEdges() += 1 :-
  max_time = MaxTime(),
  M(time: max_time, location: i),
  E(i, j);

Done() :- ActiveEdges() is null;

# M(x) :-
#   M(i),
#   targets = List{j :- M(i, k)


# G(i, j) :- E(i, j);
# G(i, i) :- M(i);

```

```python
%%logica E, W, Win, Loss, Draw

@Engine("duckdb");

E("a", "b");
E("c", "b");
E("d", "c");
E("c", "d");

W(a, b) :-
  E(a, b),
  E(b, c) => W(c, d);

Win(a) :- W(a, b);
Loss(b) :- W(a, b);
Draw(x) :- x in [a, b], E(a, b), ~W(x, y), ~W(y, x);
```

```python
%%logica Q, Move, W, Win, Loss, Draw

@Engine("duckdb");

Move("b", "a");
Move("c", "b");
Move("d", "c");
Move("c", "d");

W(a, b) :-
  Move(a, b),
  Move(b, c) => W(c, d);

Q() = if true then 5 else 7;

Node(x) distinct :- x in [a,b], Move(a,b);

Win(a) :- W(a, b);
Loss(b) :- W(a, b);
Draw(x) distinct :- Node(x), ~W(x, y), ~W(y, x);
```

```python
%%logica A

E(1, 2, t: 10);
E(2, 3, t: 5);
E(2, 3, t: 15);
E(3, 4, t: 12);
E(3, 5, t: 16);
E(3, 4, t: 20);

Start() = 1;

@OrderBy(A, "col0");

A(Start()) Min= 0;
A(b) Min= Greatest(a_t, t) :-
  a_t = A(a),
  E(a, b, t:),
  a_t < t;
```

```python
%%logica ShowEV2NodeColor, ShowEV2, E, Path, Arrival, ShowE0, ShowE

@Engine("duckdb");

N() = 20;
M() = 20;
T() = 100;

NodeName(i) = Substr("ABCDEFGHIJKLMNOPQRSTUVWXYZ", i + 1, 1);

ED(NodeName(a), NodeName(b), t0:, t1:) :-
  i in Range(M()),
  s = ToString(i),
  a = NaturalHash("a-" ++ s) % N(),
  b = NaturalHash("b-" ++ s) % N(),
  ts = NaturalHash("t0-" ++ s) % T(),
  te = NaturalHash("t1-" ++ s) % T(),
  t0 = Least(ts, te),
  t1 = Greatest(ts, te);

@OrderBy(E, "col0", "col1");
E(a, b, t0:, t1:) :- ED(a, b, t0:, t1:);
E(b, a, t0:, t1:) :- ED(a, b, t0:, t1:);

Start() = "A";
Fastest(x) = ArgMin(x -> e.t) :- e = x[Size(x) - 1];
Path(Start()) Fastest= [{a: null, b: Start(), t: 0}];
Path(b) Fastest= p ++ [{a:, b:, t:}] :-
  E(a, b, t0:, t1:),
  p = Path(a),
  pe = p[Size(p) - 1],
  t = Greatest(t0, pe.t),
  t < t1;


Arrival(Start()) Min= 0;

Arrival(b) Min= Greatest(Arrival(a), t0) :-
  E(a, b, t0:, t1:), Arrival(a) <= t1;

# @OrderBy(Arrival, "col0");
# Arrival(a) = t :-
#   p = Path(a),
#   pe = p[Size(p) - 1],
#   t = pe.t;


ShowE(a_t, b_t, label: ToString(t0) ++ "-" ++ ToString(t1)) :-
  ED(a, b, t0:, t1:),
  at = Arrival(a),
  bt = Arrival(b),
  a_t = ToString(a) ++ "@" ++ToString(ToInt64(at)),
  b_t = ToString(b) ++ "@" ++ToString(ToInt64(bt));

ShowEV2(a, b, label: ToString(t0) ++ "-" ++ ToString(t1), arrows: "", color: "#555", dashes: false) :-
  ED(a, b, t0:, t1:),
  at = Arrival(a),
  bt = Arrival(b),
  a_t = ToString(a) ++ "@" ++ToString(ToInt64(at)),
  b_t = ToString(b) ++ "@" ++ToString(ToInt64(bt));

ShowEV2(ToString(Arrival(a)), ToString(a), label: "", arrows: "to", color: "#883", dashes: true);

ShowEV2NodeColor(a) = (if Arrival(a) == 0 then
                         "rgba(160, 250, 190)"
                        else
                         "rgba(160, 190, 250)");

ShowEV2NodeColor(ToString(Arrival(a))) = "#ffa";


ShowE0(a_t, b_t, label: ToString(t0) ++ "-" ++ ToString(t1)) :-
  ED(a, b, t0:, t1:),
  at = Arrival(a),
  bt = Arrival(b),
  a_t = ToString(a),
  b_t = ToString(b);

```

```python
%%logica E, R, TC, TR

@Engine("duckdb");

N() = 20;
M() = 200;

E(a, b) distinct :-
  i in Range(M()),
  s = ToString(i),
  ax = NaturalHash("a-" ++ s) % N(),
  bx = NaturalHash("b-" ++ s) % N(),
  ax != bx,
  a = Least(ax, bx),
  b = Greatest(ax, bx);

TC(a, b) distinct :- E(a, b);
TC(a, b) distinct :- TC(a, x), TC(x, b);

TR(a, b) :- E(a, b), ~(E(a, x), TC(x, b));

R(a, b,
  arrows: "to",
  color? Max= "rgba(40, 40, 40, 0.5)",
  dashes? Min= true, width? Max= 2,
  physics? Max= false, smooth? Max= false) distinct :- E(a, b);
R(a, b,
  arrows: "to",
  color? Max= "rgba(90, 30, 30, 1.0)",
  dashes? Min= false, width? Max= 4,
  physics? Max= true, smooth? Max= true) distinct :- TR(a, b);
```

```python
from logica.common import graph
graph.SimpleGraph(R, extra_edges_columns=["arrows", "physics", "dashes", "smooth"], edge_color_column='color', edge_width_column='width', options={'width': 1000})
```

```python
from logica.common import graph
graph.DirectedGraph(TR)
```

```python
%%logica E, TC, CC, ECC, Render

@Engine("duckdb");

N() = 20;
M() = 40;

E(a, b) distinct :-
  i in Range(M()),
  s = ToString(i),
  a = NaturalHash("a---" ++ s) % N(),
  b = NaturalHash("b---" ++ s) % N(),
  a != b;

Node(x) :- x in [a,b], E(a,b);

TC(a, b) distinct :- E(a, b);
TC(a, b) distinct :- TC(a, x), TC(x, b);

CC(x) Min= x :- Node(x);
CC(x) Min= y :- TC(x, y), TC(y, x);

ECC(CC(a), CC(b)) distinct :- E(a, b);

NodeName(x) = ToString(ToInt64(x));
CompName(x) = "c-" ++ ToString(ToInt64(x));

Render(NodeName(a), NodeName(b), physics: true, arrows: "to", dashes: false, smooth: true, color: "#33e") :- E(a, b);
Render(CompName(x), CompName(y), physics: true, arrows: "to", dashes: false, smooth: true, color: "#33e") :- ECC(x, y);
Render(NodeName(ToInt64(a)), CompName(CC(a)), physics: false, arrows: "to", dashes: true, smooth: false, color: "#888");
```

```python
graph.DirectedGraph(E)
```

```python
graph.SimpleGraph(Render, extra_edges_columns=['arrows', 'physics', 'dashes', 'smooth'], edge_color_column='color', options={'width': 1000, 'height': 1000})
```
